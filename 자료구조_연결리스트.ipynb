# 1. 단일연결리스트

```python
class 노드:
    def __init__(self, value):
        self.값 = value
        self.다음 = None

class 단일연결리스트:
    def __init__(self):
        self.머리노드 = None

리스트 = 단일연결리스트()

첫번째노드 = 노드(10)
두번째노드 = 노드(20)
세번째노드 = 노드(30)

리스트.머리노드 = 첫번째노드

리스트.머리노드.다음 = 두번째노드

두번째노드.다음 = 세번째노드

current = 리스트.머리노드
for i in range(3):
    print(current.값)
    current = current.다음
```

```python
class Node: # 노드 객체를 만들기 위함.
    def __init__(self, item):
        self.item = item # 노드가 가지고 있는 값
        self.next = None # 다음 노드를 담기 위한 변수


class SingleLinkedList: # 노드 객체들을 담기위한 객체
    def __init__(self):
        self.head = None # 단일 연결리스트에서 어떤 노드가 맨 앞노드인지



linked_list = SingleLinkedList() # 단일 연결리스트 생성

linked_list.head = Node(1) # 단일 연결리스트에서 맨 앞 노드를 1의 값을 가지고 있는 노드로 설정
second = Node(2) # 2의 값을 가지고 있는 노드를 생성
third = Node(3) # 3의 값을 가지고 있는 노드를 생성

linked_list.head.next = second # 맨 앞 노드(첫번째)으 다음 노드는 2의 값을 가지고 있는 노드로 설정(두번째 노드)
second.next = third # 두번째 노드의 다음 노드는 3의 값을 가지고 있는 노드

while linked_list.head != None: # 단일 연결 리스트에서 맨 앞노드가 있을때 까지.
    print(linked_list.head.item, end=" ")
    linked_list.head = linked_list.head.next
```

# 2. 이중연결리스트

```python
class Node: # 노드 객체를 만들기 위함.
    def __init__(self, item):
        self.item = item # 노드가 가지고 있는 값
        self.prev = None
        self.next = None # 다음 노드를 담기 위한 변수

class DoubleLinkedList: # 노드 객체들을 담기위한 객체
    def __init__(self):
        self.head = None # 이중 연결리스트에서 어떤 노드가 맨 앞노드인지

Double_linked_list = DoubleLinkedList() # 이중 연결리스트 생성

first_node = Node(1) # 1의 값을 가지고 있는 노드를 생성
second_node = Node(2) # 2의 값을 가지고 있는 노드를 생성
Double_linked_list.head = first_node # 이중 연결리스트에서 맨 앞 노드를 1의 값을 가지고 있는 노드로 설정
Double_linked_list.head.next = second_node # 맨 앞 노드(첫번째)으 다음 노드는 2의 값을 가지고 있는 노드로 설정(두번째 노드)
second_node.prev = first_node

while Double_linked_list.head != None: # 이중 연결 리스트에서 맨 앞노드가 있을때 까지.
    print(Double_linked_list.head.item, end=" ")
    Double_linked_list.head = Double_linked_list.head.next
```

# 3. 원형연결리스트

```python
class Node:
    def __init__(self,item):
        self.item = item # 노드가 가지고 있는 값
        self.next = None

class CircularLinkedList: # 노드 객체들을 담기위한 객체
    def __init__(self):
        self.head = None # 원형 연결리스트에서 어떤 노드가 맨 앞노드인지

```

```python
Circular_linked_list = CircularLinkedList() # 원형 연결리스트 생성

# 노드생성
one = Node(3)
two = Node(5)
three = Node(9)

# 노드연결
Circular_linked_list.head = one
one.next = two
two.next = three
three.next = one


while Circular_linked_list.head != None: # 끝없이 돌 것이다 (원형이기 때문)
    print(Circular_linked_list.head.item, end=" ")
    Circular_linked_list.head = Circular_linked_list.head.next
```
