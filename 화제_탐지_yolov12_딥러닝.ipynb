```python
from sahi.predict import get_sliced_prediction
from sahi.models.yolov8 import Yolov8DetectionModel
import os

# YOLOv12 학습 완료 모델 불러오기
detection_model = Yolov8DetectionModel(
    model_path="/content/runs/detect/fire_yolov122/weights/best.pt",
    confidence_threshold=0.3,
    device="cuda"
)

# 예측할 이미지 경로
image_path = "/content/runs/detect/fire_yolov122/train_batch0.jpg"  # 실제 이미지 파일명으로 교체

# SAHI 적용 (작은 객체 감지 강화)
result = get_sliced_prediction(
    image=image_path,
    detection_model=detection_model,
    slice_height=512,   # 잘라서 예측할 타일 크기
    slice_width=512,
    overlap_height_ratio=0.2,
    overlap_width_ratio=0.2
)

# 결과 저장 및 시각화
os.makedirs("/content/sahi_results", exist_ok=True)
result.export_visuals(export_dir="/content/sahi_results/")
print("SAHI 결과 저장 완료 ✅ /content/sahi_results/")
```
